1. Em uma pesquisa sobre a relação da prática de atividade física no colesterol total de indivíduos saudáveis, o que se pode afirmar sobre as populações investigadas com os dados amostrais a seguir?




```python
#@title **Vamos importar a biblioteca SciPy, que oferece uma interface robusta para distribuições.**

from scipy.stats import norm
import math
```

```python

# Valor crítico para intervalo de confiança de 95% (z*)
z_star_95 = norm.ppf(0.975)

# Praticando atividade física
mean_p = 200
std_dev_p = 4.93
n_p = 100
margin_of_error_p = z_star_95 * (std_dev_p / math.sqrt(n_p))

lower_p = mean_p - margin_of_error_p
upper_p = mean_p + margin_of_error_p

# Não praticando atividade física
mean_n = 209.8
std_dev_n = 4.95
n_n = 100
margin_of_error_n = z_star_95 * (std_dev_n / math.sqrt(n_n))

lower_n = mean_n - margin_of_error_n
upper_n = mean_n + margin_of_error_n

print("Intervalo de confiança para praticantes de atividade física:")
print(f"Limite inferior: {lower_p:.2f}, Limite superior: {upper_p:.2f}")

print("\nIntervalo de confiança para não praticantes de atividade física:")
print(f"Limite inferior: {lower_n:.2f}, Limite superior: {upper_n:.2f}")
```

# Intervalo de confiança - Média usando a distribuição T

Se a média de IMC (Kg/m2) de uma determinada amostra de tamanho 30 é 25,4 kg/m2 e o desvio-padrão é 6,0 kg/m2 . O que se pode afirmar sobre o IMC médio da população (μ) a um nível de 95% de confiança?
Supondo o desvio padrão populacional desconhecido, calcule o IC.


```python
from scipy.stats import t
import math

# Valor crítico para intervalo de confiança de 95% com 29 graus de liberdade (z*)
z_star_95 = t.ppf(0.975, df=29)

# Cálculo do intervalo de confiança
media = 25.4
std_dev = 6
n = 30

margin_of_error = z_star_95 * (std_dev / math.sqrt(n))

lower = media - margin_of_error
upper = media + margin_of_error

print("Intervalo de confiança:")
print(f"Limite inferior: {lower:.2f}, Limite superior: {upper:.2f}")
```
